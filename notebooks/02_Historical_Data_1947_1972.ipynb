# Historical INR/USD Exchange Rate (1947–1972)

## Data Source and Methodology

The Reserve Bank of India (RBI) provides historical USD-INR exchange-rate information for 1945–1971.

The historical data is reported in rupees per 100 US dollars for the earlier financial years. To make the figures comparable with the FRED dataset, we convert these values into rupees per 1 US dollar by dividing by 100.

The historical records use financial-year averages, whereas the FRED dataset contains monthly observations. These differences in frequency and reporting periods must be considered when combining or comparing the datasets.

```python
import pandas as pd

historical_data = {
    "Financial_Year": [
        "1947-48", "1948-49", "1949-50",
        "1950-51", "1951-52", "1952-53",
        "1953-54", "1954-55", "1955-56",
        "1956-57", "1957-58", "1958-59",
        "1959-60", "1960-61", "1961-62",
        "1962-63", "1963-64"
    ],
    "RBI_INR_per_100_USD": [
        331.75, 331.75, 407.4375,
        477.50, 478.15625, 478.25,
        476.25, 477.78125, 478.9375,
        479.22, 478.32, 476.54,
        476.80, 476.82, 477.20,
        477.20, 478.29
    ]
}

historical_df = pd.DataFrame(historical_data)

historical_df["INR_per_USD"] = (
    historical_df["RBI_INR_per_100_USD"] / 100
)

display(historical_df)
```

```python
historical_df.to_csv(
    "../data/raw/inr_usd_historical_1947_1964.csv",
    index=False
)

print("RBI historical data saved successfully!")
```

```python
print("Rows:", len(historical_df))
print("Columns:", historical_df.columns.tolist())

display(historical_df.head())
display(historical_df.tail())

print("Minimum rate:", historical_df["INR_per_USD"].min())
print("Maximum rate:", historical_df["INR_per_USD"].max())
```

```python
later_data = pd.DataFrame({
    "Financial_Year": [
        "1965-66", "1966-67",
        "1966-67", "1967-68",
        "1968-69", "1969-70", "1970-71"
    ],
    "USD_per_INR_100": [
        20.90, 20.86, 13.20,
        13.22, 13.12, 13.26, 13.23
    ],
    "Regime": [
        "Before devaluation",
        "Before devaluation",
        "After devaluation",
        "After devaluation",
        "After devaluation",
        "After devaluation",
        "After devaluation"
    ]
})

later_data["INR_per_USD"] = (
    100 / later_data["USD_per_INR_100"]
)

display(later_data.round(4))
```

## Historical Event: 1966 Rupee Devaluation

On June 6, 1966, India devalued the rupee from approximately ₹4.76 to ₹7.50 per US dollar.

The IMF records this as a 36.5% devaluation in the rupee's external value.

The adjustment was intended to improve export competitiveness and address external economic pressures.

This event represents a major structural change in India's exchange-rate history. It should be distinguished from gradual market-driven depreciation in later decades.

**Source:** International Monetary Fund, *Other Adjustments in Exchange Rates (1966–70)*.


## 1971–1972: Transition to a Floating Exchange Rate

India's exchange-rate system underwent another important transition between 1971 and 1972.

- Before mid-December 1971, the official exchange rate was ₹7.50 per US dollar.
- From mid-December 1971 to the end of June 1972, it changed to ₹7.27927 per US dollar.
- After June 1972, India moved to a floating exchange-rate system.

The annual average exchange rates were ₹7.444 for 1971–72 and ₹7.706 for 1972–73.

These figures show why it is important to distinguish between official rates at specific dates and annual average exchange rates.

**Source:** World Bank historical currency-equivalent tables, citing IMF International Financial Statistics.


```python
transition_data = pd.DataFrame({
    "Period": [
        "Before mid-December 1971",
        "Mid-December 1971 to June 1972",
        "After June 1972"
    ],
    "INR_per_USD": [
        7.50,
        7.27927,
        None
    ],
    "Exchange_Rate_System": [
        "Official fixed rate",
        "Adjusted official rate",
        "Floating exchange rate"
    ]
})

annual_rates = pd.DataFrame({
    "Financial_Year": ["1971-72", "1972-73"],
    "Annual_Average_INR_per_USD": [7.444, 7.706]
})

display(transition_data)
display(annual_rates)
```

```python
transition_data.to_csv(
    "../data/raw/inr_usd_transition_1971_1972.csv",
    index=False
)

annual_rates.to_csv(
    "../data/raw/inr_usd_annual_average_1971_1973.csv",
    index=False
)

print("1971–1972 transition data saved!")
```

```python
# Combine RBI historical observations
historical_combined = pd.concat(
    [
        historical_df[
            ["Financial_Year", "INR_per_USD"]
        ],
        later_data[
            ["Financial_Year", "INR_per_USD"]
        ]
    ],
    ignore_index=True
)

print("Historical observations:", len(historical_combined))
display(historical_combined)

print("\nExchange-rate transition:")
display(annual_rates)
```
